```python
import os
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
```

```python
DATASET_PATH = Path("../dataset")
```

```python
dataset_candidates = [
    Path("../dataset").resolve(),
    Path.cwd() / "dataset",
    Path.cwd().parent / "dataset",
]

DATASET_PATH = next((p for p in dataset_candidates if p.exists()), dataset_candidates[0])

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset directory not found. Checked: {[str(p) for p in dataset_candidates]}"
    )

crops = [folder.name for folder in DATASET_PATH.iterdir() if folder.is_dir()]

print("Total Crops:", len(crops))
print(crops)
```

```python
disease_classes = []

for crop in DATASET_PATH.iterdir():

    train_folder = crop / "Train"

    if train_folder.exists():

        for disease in train_folder.iterdir():

            if disease.is_dir():

                disease_classes.append(
                    f"{crop.name} - {disease.name}"
                )

print("Total Classes:", len(disease_classes))
```

```python
class_counts = []

for crop in DATASET_PATH.iterdir():

    train_folder = crop / "Train"

    if train_folder.exists():

        for disease in train_folder.iterdir():

            if disease.is_dir():

                image_count = len(list(disease.glob("*")))

                class_counts.append(
                    {
                        "Crop": crop.name,
                        "Disease": disease.name,
                        "Train Images": image_count
                    }
                )

df = pd.DataFrame(class_counts)

df.head()
```

```python
print(df.sort_values(
    by="Train Images",
    ascending=False
).head())

print()

print(df.sort_values(
    by="Train Images"
).head())
```

```python
plt.figure(figsize=(14,6))

plt.bar(
    df["Disease"],
    df["Train Images"]
)

plt.xticks(rotation=90)

plt.title("Training Images Per Disease Class")

plt.tight_layout()

plt.savefig(
    "../results/class_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
