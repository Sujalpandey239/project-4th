```python
import numpy as np

import faiss
```

```python
embeddings = np.load(
    "../embeddings/embeddings.npy"
)

print(
    embeddings.shape
)
```

```python
embeddings = embeddings.astype(
    "float32"
)
```

```python
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(
    dimension
)

index.add(
    embeddings
)

print(
    "Vectors in index:",
    index.ntotal
)
```

```python
faiss.write_index(
    index,
    "../faiss_index/plant_disease.index"
)
```
